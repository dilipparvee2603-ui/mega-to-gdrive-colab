# 🚀 MEGA to Google Drive — Full Web GUI
A complete web-based control center running on Google Colab's high-speed cloud infrastructure to transfer files directly from MEGA to Google Drive.

### ✨ Key Features:
- **Full Web Interface**: Accessible in Colab or via a shareable public URL on your browser/phone.
- **Movie & Subfolder Grouping**: Download multiple links (e.g. CD1 & CD2 or seasons/episodes) into dedicated folders automatically.
- **One-by-One Queue Builder**: Add, review, and manage download links in a visual table before starting.
- **Real-time Live Progress**: Live terminal streaming, download speeds, and completion indicators.
- **Drive Storage Monitor**: Visual gauge tracking remaining storage on your Google Drive.

### Step 2: Launch Full Web GUI
Run this cell to launch the Web Application. You will receive an inline dashboard right here in Colab, as well as a **public `.gradio.live` link** that you can open in any browser or on your phone!

```python
#@title Launch Web Control Center { display-mode: "form" }
# Set SHARE_URL = True to generate a public link accessible from any browser or phone
SHARE_URL = True #@param {type:"boolean"}

import os
import sys
import re
import time
import shutil
import subprocess
import pandas as pd
import gradio as gr

# Optional PTY support on Linux (Colab)
try:
    import pty
    import fcntl
    HAS_PTY = True
except ImportError:
    HAS_PTY = False

# Global Queue State: [{'id': 1, 'subfolder': '...', 'link': '...', 'status': 'Pending'}]
if 'queue_state' not in globals():
    queue_state = []

def extract_mega_links(text):
    if not text:
        return []
    pattern = r'(?:https?://)?(?:www\.)?mega\.(?:nz|io|co\.nz)/(?:file/|folder/|embed/|#)[^\s,"\'<>]+'
    found = re.findall(pattern, text)
    if not found:
        candidates = [line.strip() for line in re.split(r'[\r\n,]+', text) if line.strip()]
        found = [c for c in candidates if "mega." in c or c.startswith("http")]
    
    clean_links = []
    for link in found:
        link = link.strip().rstrip('.,;')
        if link and not link.startswith("http://") and not link.startswith("https://"):
            link = "https://" + link
        if link and link not in clean_links:
            clean_links.append(link)
    return clean_links

def get_queue_df():
    if not queue_state:
        return pd.DataFrame(columns=["#", "Movie / Subfolder", "MEGA Link", "Status"])
    rows = []
    for item in queue_state:
        sub = item["subfolder"] if item["subfolder"] else "(Base Folder)"
        rows.append({
            "#": item["id"],
            "Movie / Subfolder": sub,
            "MEGA Link": item["link"],
            "Status": item["status"]
        })
    return pd.DataFrame(rows)

def add_to_queue(subfolder, links_text):
    global queue_state
    if not links_text or not links_text.strip():
        return get_queue_df(), "", "⚠️ Please paste at least one MEGA link."
    
    # Support input as a text file path
    if os.path.isfile(links_text.strip()):
        try:
            with open(links_text.strip(), 'r', encoding='utf-8', errors='ignore') as f:
                links_text = f.read()
        except Exception as e:
            return get_queue_df(), "", f"❌ Failed to read file: {e}"
            
    clean_sub = re.sub(r'[<>:"/\\|?*]', '_', subfolder.strip()).strip(' .') if subfolder else ""
    links = extract_mega_links(links_text)
    
    if not links:
        return get_queue_df(), links_text, "❌ No valid MEGA links found. Format: https://mega.nz/file/... or /folder/..."
    
    added = 0
    for link in links:
        if not any(item["link"] == link for item in queue_state):
            queue_state.append({
                "id": len(queue_state) + 1,
                "subfolder": clean_sub,
                "link": link,
                "status": "Pending"
            })
            added += 1
            
    target_label = f"folder '{clean_sub}'" if clean_sub else "base folder"
    msg = f"✅ Added {added} link(s) to {target_label}. Total queued: {len(queue_state)}"
    return get_queue_df(), "", msg

def remove_last():
    global queue_state
    if queue_state:
        removed = queue_state.pop()
        msg = f"ℹ️ Removed #{removed['id']}: {removed['link']}"
    else:
        msg = "ℹ️ Queue is already empty."
    return get_queue_df(), msg

def clear_queue():
    global queue_state
    queue_state = []
    return get_queue_df(), "ℹ️ Queue cleared."

def get_storage_info():
    path = "/content/drive/MyDrive"
    if not os.path.exists(path):
        return "⚠️ Google Drive not mounted at /content/drive/MyDrive. Please run Step 1."
    try:
        total, used, free = shutil.disk_usage(path)
        total_gb = total / (1024 ** 3)
        used_gb = used / (1024 ** 3)
        free_gb = free / (1024 ** 3)
        pct = (used / total) * 100
        return f"📁 **Google Drive Storage:**\n- **Free Space:** {free_gb:.2f} GB\n- **Used Space:** {used_gb:.2f} GB ({pct:.1f}%)\n- **Total Capacity:** {total_gb:.2f} GB"
    except Exception as e:
        return f"⚠️ Unable to read storage info: {e}"

def run_transfers(base_folder):
    global queue_state
    if not queue_state:
        yield get_queue_df(), "⚠️ Queue is empty! Add at least one link before starting.", "⚠️ Queue is empty.", gr.update(interactive=True)
        return
        
    base_folder = base_folder.strip() or "MEGA_Transfers"
    base_target = f"/content/drive/MyDrive/{base_folder}"
    os.makedirs(base_target, exist_ok=True)
    
    # Disable start button
    yield get_queue_df(), "🚀 Initializing MEGA transfer engine...\n", "⏳ Transfers in progress...", gr.update(interactive=False)
    
    subprocess.run(["mega-quit"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    time.sleep(1)
    
    log_history = f"🚀 Target Folder: Google Drive -> {base_folder}\nTotal items to download: {len(queue_state)}\n" + ("=" * 60) + "\n"
    yield get_queue_df(), log_history, "⏳ Transfers in progress...", gr.update(interactive=False)
    
    success_count = 0
    fail_count = 0
    
    for idx, item in enumerate(queue_state, 1):
        if item["status"] == "Completed ✅":
            continue
            
        item["status"] = "Transferring ⏳"
        yield get_queue_df(), log_history, f"⏳ [{idx}/{len(queue_state)}] Transferring...", gr.update(interactive=False)
        
        sub = item.get("subfolder", "").strip()
        if sub:
            dest = os.path.join(base_target, sub)
            dest_label = f"{base_folder} / {sub}"
        else:
            dest = base_target
            dest_label = base_folder
        os.makedirs(dest, exist_ok=True)
        
        log_history += f"\n[TRANSFER {idx}/{len(queue_state)}]\nDestination: Google Drive -> {dest_label}\nLink: {item['link']}\n" + ("-" * 50) + "\n"
        yield get_queue_df(), log_history, f"⏳ [{idx}/{len(queue_state)}] Downloading to {dest_label}...", gr.update(interactive=False)
        
        cmd = ["mega-get", item["link"], dest]
        
        if HAS_PTY:
            master, slave = pty.openpty()
            process = subprocess.Popen(cmd, stdout=slave, stderr=slave, text=True, close_fds=True)
            os.close(slave)
            
            fl = fcntl.fcntl(master, fcntl.F_GETFL)
            fcntl.fcntl(master, fcntl.F_SETFL, fl | os.O_NONBLOCK)
            
            last_yield = time.time()
            while True:
                try:
                    chunk = os.read(master, 1024)
                    if chunk:
                        log_history += chunk.decode('utf-8', errors='ignore')
                        if time.time() - last_yield > 0.35:
                            yield get_queue_df(), log_history, f"⏳ [{idx}/{len(queue_state)}] Downloading...", gr.update(interactive=False)
                            last_yield = time.time()
                except OSError:
                    pass
                    
                if process.poll() is not None:
                    try:
                        rem = os.read(master, 4096)
                        if rem:
                            log_history += rem.decode('utf-8', errors='ignore')
                    except:
                        pass
                    break
                time.sleep(0.1)
                
            try:
                os.close(master)
            except:
                pass
        else:
            process = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
            for line in process.stdout:
                log_history += line
                yield get_queue_df(), log_history, f"⏳ [{idx}/{len(queue_state)}] Downloading...", gr.update(interactive=False)
            process.wait()
            
        if process.poll() is None:
            process.terminate()
            process.wait()
            
        if process.returncode == 0:
            item["status"] = "Completed ✅"
            success_count += 1
            log_history += f"\n✅ Successfully downloaded into '{dest_label}'\n"
        else:
            item["status"] = "Failed ❌"
            fail_count += 1
            log_history += f"\n❌ Transfer failed with exit code {process.returncode}\n"
            
        yield get_queue_df(), log_history, f"Status: {idx}/{len(queue_state)} done", gr.update(interactive=False)
        
    subprocess.run(["mega-quit"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    
    log_history += "\n" + ("=" * 60) + f"\n🎉 All finished! Successful: {success_count} | Failed: {fail_count}\n"
    summary_msg = f"🎉 Session Complete: {success_count}/{len(queue_state)} succeeded."
    if fail_count > 0:
        summary_msg += f" ({fail_count} failed)"
        
    yield get_queue_df(), log_history, summary_msg, gr.update(interactive=True)

# Build the Gradio Interface with custom theme & tabs
custom_css = """
#title_header { text-align: center; margin-bottom: 8px; }
.folder_box { background-color: #f8f9fa; border-radius: 8px; padding: 10px; }
#start_button { background: linear-gradient(90deg, #10b981 0%, #059669 100%); color: white; font-weight: bold; font-size: 16px; border: none; }
#start_button:hover { background: #047857; }
"""

with gr.Blocks(title="MEGA to Google Drive Cloud Station", css=custom_css, theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🚀 MEGA to Google Drive Cloud Transfer Station", elem_id="title_header")
    gr.Markdown("Fast, cloud-direct transfer between MEGA and your personal Google Drive without consuming local bandwidth.", elem_id="title_header")
    
    with gr.Tabs():
        with gr.TabItem("📥 Queue & Transfer"):
            with gr.Row():
                base_folder_input = gr.Textbox(
                    label="Base Google Drive Folder",
                    value="MEGA_Transfers",
                    placeholder="Folder name in MyDrive"
                )
                subfolder_input = gr.Textbox(
                    label="Movie / Subfolder Name (Optional)",
                    placeholder="e.g. Deadpool and Wolverine (2024) or Part1-Part2",
                    info="Links added with this name will download into this dedicated subfolder"
                )
                
            links_input = gr.Textbox(
                label="MEGA Link(s)",
                placeholder="Paste MEGA link here (or multiple links separated by commas/newlines, or path to links.txt)...",
                lines=2
            )
            
            with gr.Row():
                add_btn = gr.Button("➕ Add to Queue", variant="primary")
                remove_btn = gr.Button("➖ Remove Last", variant="secondary")
                clear_btn = gr.Button("🗑️ Clear Queue", variant="stop")
                
            status_notice = gr.Markdown("💡 *Queue is ready. Add links above to build your transfer list.*")
            
            gr.Markdown("### 📋 Current Transfer Queue")
            queue_table = gr.Dataframe(
                value=get_queue_df(),
                headers=["#", "Movie / Subfolder", "MEGA Link", "Status"],
                datatype=["number", "str", "str", "str"],
                interactive=False,
                wrap=True
            )
            
            start_btn = gr.Button("🚀 Start Transfers", elem_id="start_button", size="lg")
            
            gr.Markdown("### 🖥️ Live Transfer Terminal Output")
            log_box = gr.Textbox(
                label="Real-time Progress Console",
                value="Waiting to start...\n",
                lines=10,
                max_lines=25,
                autoscroll=True,
                interactive=False
            )
            
        with gr.TabItem("📊 Google Drive Storage & Tools"):
            storage_markdown = gr.Markdown(value=get_storage_info())
            refresh_storage_btn = gr.Button("🔄 Refresh Storage Usage")
            
            gr.Markdown("---")
            gr.Markdown("### ⚡ Google Drive Tools")
            sync_info = gr.Markdown("")
            sync_btn = gr.Button("🔄 Flush Drive Cache (Force Instant Sync to Phone/PC)")
            
            def flush_cache():
                try:
                    from google.colab import drive
                    drive.flush_and_unmount()
                    return "✅ Cache flushed and unmounted successfully! Files are now synced on Google Drive."
                except Exception as e:
                    return f"⚠️ Cache flush notice: {e}"
                    
            sync_btn.click(flush_cache, outputs=[sync_info])
            refresh_storage_btn.click(get_storage_info, outputs=[storage_markdown])
            
    # Wire button event listeners
    add_btn.click(
        fn=add_to_queue,
        inputs=[subfolder_input, links_input],
        outputs=[queue_table, links_input, status_notice]
    )
    
    remove_btn.click(
        fn=remove_last,
        outputs=[queue_table, status_notice]
    )
    
    clear_btn.click(
        fn=clear_queue,
        outputs=[queue_table, status_notice]
    )
    
    start_btn.click(
        fn=run_transfers,
        inputs=[base_folder_input],
        outputs=[queue_table, log_box, status_notice, start_btn]
    )

print("\n\033[94m[INFO] Starting Gradio Web Server...\033[0m")
demo.queue().launch(share=SHARE_URL, debug=False)
```
